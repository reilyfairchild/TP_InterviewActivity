#### Key Takeaways:

- Figure 1: Prior to the joining the workforce training program, the median annual earning was ~\$12,000/year, rising to ~\$25,000/year by year 3 of the program, **a 108% increase in annual earnings**. This indicates the program supports upward mobility in job opportunities and earnings, as clients continued to see notable wage increases 3 years after the program. 
- Figure 2: Participants across all race/ethnicity groups showed over a **~\$10,000 increase in annual earnings** after the program. Each racial/ethnic group showed similar increases in earnings, indicating the training program did not prioritize/benefit one group more than another.
- Figure 3: Similarly, each education attainment group's **annual earnings increased roughly \$10,000 after the program.** This highlights the program's diversity in training opportunities across varying education levels.
